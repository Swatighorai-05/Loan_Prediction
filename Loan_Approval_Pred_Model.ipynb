## 1. Model Evaluation

Evaluating the baseline **Logistic Regression** model (`model`) already trained in this notebook using standard classification metrics: **Accuracy**, **Precision**, **Recall**, and **F1-score**.

Since `loan_status` is a **binary classification problem** (`1: Approved`, `0: Rejected`), metrics are computed with `average='binary'`.


```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix
import pandas as pd
import numpy as np

# Baseline Model: Logistic Regression (already trained as `model`)
y_train_pred_log = model.predict(x_train_scaled)
y_test_pred_log = model.predict(x_test_scaled)

# Calculate Classification Metrics for Test Set
acc_log = accuracy_score(y_test, y_test_pred_log)
prec_log = precision_score(y_test, y_test_pred_log, average='binary')
rec_log = recall_score(y_test, y_test_pred_log, average='binary')
f1_log = f1_score(y_test, y_test_pred_log, average='binary')

# Display Baseline Evaluation Table
baseline_metrics_df = pd.DataFrame([{
    'Model': 'Logistic Regression (Baseline)',
    'Accuracy': round(acc_log, 4),
    'Precision': round(prec_log, 4),
    'Recall': round(rec_log, 4),
    'F1-score': round(f1_log, 4)
}])

print("=== BASELINE MODEL EVALUATION (Logistic Regression) ===")
print(classification_report(y_test, y_test_pred_log, target_names=['Rejected (0)', 'Approved (1)']))
display(baseline_metrics_df)
```

## 2. Check Overfitting / Underfitting

Comparing performance on **Training Data** vs **Testing Data** to evaluate fit status.

### Criteria & Rules:
- **Possible Overfitting**: Train score is significantly higher than Test score (Difference > 0.05).
- **Possible Underfitting**: Both Train and Test scores are low (< 0.70).
- **Good Fit**: Train and Test scores are reasonably close (Difference <= 0.05) and high.


```python
train_acc_log = accuracy_score(y_train, y_train_pred_log)
test_acc_log = acc_log
diff_log = train_acc_log - test_acc_log

def get_fit_status(train_score, test_score, threshold=0.05):
    if train_score - test_score > threshold:
        return "Possible Overfitting"
    elif train_score < 0.70 and test_score < 0.70:
        return "Possible Underfitting"
    else:
        return "Good Fit"

fit_status_log = get_fit_status(train_acc_log, test_acc_log)

fit_table_log = pd.DataFrame([{
    'Model': 'Logistic Regression (Baseline)',
    'Train Score': round(train_acc_log, 4),
    'Test Score': round(test_acc_log, 4),
    'Difference': round(diff_log, 4),
    'Fit Status': fit_status_log
}])

display(fit_table_log)
```

## 3. Cross-Validation (5-Fold Stratified K-Fold)

Performing **5-Fold Stratified Cross-Validation** on the **TRAINING DATA ONLY** (`x_train_scaled`, `y_train`) to evaluate generalization performance without data leakage.

- **StratifiedKFold** ensures that class distributions (Approved vs Rejected) are maintained across all 5 folds.


```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

# Stratified 5-Fold Cross-Validation on Training Data ONLY
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores_log = cross_val_score(model, x_train_scaled, y_train, cv=skf, scoring='accuracy')

cv_mean_log = cv_scores_log.mean()
cv_std_log = cv_scores_log.std()
cv_min_log = cv_scores_log.min()
cv_max_log = cv_scores_log.max()

cv_table_log = pd.DataFrame([{
    'Model': 'Logistic Regression (Baseline)',
    'CV Mean': round(cv_mean_log, 4),
    'CV Std': round(cv_std_log, 4),
    'Minimum': round(cv_min_log, 4),
    'Maximum': round(cv_max_log, 4)
}])

print("Individual 5-Fold Accuracy Scores for Logistic Regression:", np.round(cv_scores_log, 4))
display(cv_table_log)
```

## 4. Advanced Models (Bagging & Boosting)

Training advanced classification models on the existing training data (`x_train_scaled`, `y_train`) to improve loan approval prediction:

1. **Decision Tree Classifier**
2. **Random Forest Classifier** (Bagging Ensemble)
3. **AdaBoost Classifier** (Boosting Ensemble)
4. **Gradient Boosting Classifier** (Boosting Ensemble)


```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier

# Initialize Advanced Models
dt_model = DecisionTreeClassifier(random_state=42)
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
ada_model = AdaBoostClassifier(n_estimators=100, random_state=42)
gb_model = GradientBoostingClassifier(n_estimators=100, random_state=42)

models = {
    'Logistic Regression': model,
    'Decision Tree': dt_model,
    'Random Forest (Bagging)': rf_model,
    'AdaBoost': ada_model,
    'Gradient Boosting': gb_model
}

model_results = []

for name, clf in models.items():
    # Train if not already trained
    if name != 'Logistic Regression':
        clf.fit(x_train_scaled, y_train)
    
    # Predictions
    y_tr_pred = clf.predict(x_train_scaled)
    y_te_pred = clf.predict(x_test_scaled)
    
    # Metrics
    tr_acc = accuracy_score(y_train, y_tr_pred)
    te_acc = accuracy_score(y_test, y_te_pred)
    prec = precision_score(y_test, y_te_pred, average='binary')
    rec = recall_score(y_test, y_te_pred, average='binary')
    f1 = f1_score(y_test, y_te_pred, average='binary')
    
    # 5-Fold Stratified CV on Training Data ONLY
    scores = cross_val_score(clf, x_train_scaled, y_train, cv=skf, scoring='accuracy')
    cv_mean = scores.mean()
    cv_std = scores.std()
    
    fit_status = get_fit_status(tr_acc, te_acc)
    
    model_results.append({
        'Model': name,
        'Accuracy': round(te_acc, 4),
        'Precision': round(prec, 4),
        'Recall': round(rec, 4),
        'F1-score': round(f1, 4),
        'CV Mean': round(cv_mean, 4),
        'CV Std': round(cv_std, 4),
        'Train Score': round(tr_acc, 4),
        'Test Score': round(te_acc, 4),
        'Fit Status': fit_status
    })

comparison_df = pd.DataFrame(model_results)
display(comparison_df)
```

## 5. Hyperparameter Tuning

Applying **GridSearchCV** on **TRAINING DATA ONLY** (`x_train_scaled`, `y_train`) to optimize hyperparameters for:
- **Logistic Regression** (`C`, `solver`)
- **Random Forest Classifier** (`n_estimators`, `max_depth`, `min_samples_split`, `min_samples_leaf`)


```python
from sklearn.model_selection import GridSearchCV

print("=== 1. HYPERPARAMETER TUNING: LOGISTIC REGRESSION ===")
param_grid_log = {
    'C': [0.01, 0.1, 1.0, 10.0],
    'solver': ['lbfgs', 'liblinear']
}
grid_log = GridSearchCV(LogisticRegression(random_state=42), param_grid_log, cv=skf, scoring='accuracy')
grid_log.fit(x_train_scaled, y_train)

best_log_model = grid_log.best_estimator_
tuned_log_test_acc = accuracy_score(y_test, best_log_model.predict(x_test_scaled))

print(f"Best Parameters (Logistic Regression): {grid_log.best_params_}")
print(f"Best CV Score (Logistic Regression): {grid_log.best_score_:.4f}")
print(f"Test Accuracy (Tuned Logistic Regression): {tuned_log_test_acc:.4f}")

print("\n=== 2. HYPERPARAMETER TUNING: RANDOM FOREST ===")
param_grid_rf = {
    'n_estimators': [50, 100, 150],
    'max_depth': [10, 15, 20, None],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}
grid_rf = GridSearchCV(RandomForestClassifier(random_state=42), param_grid_rf, cv=skf, scoring='accuracy')
grid_rf.fit(x_train_scaled, y_train)

best_rf_model = grid_rf.best_estimator_
tuned_rf_test_acc = accuracy_score(y_test, best_rf_model.predict(x_test_scaled))

print(f"Best Parameters (Random Forest): {grid_rf.best_params_}")
print(f"Best CV Score (Random Forest): {grid_rf.best_score_:.4f}")
print(f"Test Accuracy (Tuned Random Forest): {tuned_rf_test_acc:.4f}")

# Hyperparameter Tuning Improvement Table
tuning_comparison = pd.DataFrame([
    {
        'Model': 'Logistic Regression',
        'Original Test Score': round(acc_log, 4),
        'Tuned Test Score': round(tuned_log_test_acc, 4),
        'Improvement': round(tuned_log_test_acc - acc_log, 4)
    },
    {
        'Model': 'Random Forest',
        'Original Test Score': round(accuracy_score(y_test, rf_model.predict(x_test_scaled)), 4),
        'Tuned Test Score': round(tuned_rf_test_acc, 4),
        'Improvement': round(tuned_rf_test_acc - accuracy_score(y_test, rf_model.predict(x_test_scaled)), 4)
    }
])

display(tuning_comparison)
```

## 6. Compare All Models

Consolidating final metrics for baseline, advanced, and hyperparameter-tuned models into a master comparison table.


```python
all_final_models = {
    'Logistic Regression (Baseline)': model,
    'Decision Tree': dt_model,
    'Random Forest (Baseline)': rf_model,
    'AdaBoost': ada_model,
    'Gradient Boosting': gb_model,
    'Tuned Logistic Regression': best_log_model,
    'Tuned Random Forest': best_rf_model
}

final_results = []
for name, clf in all_final_models.items():
    y_tr_p = clf.predict(x_train_scaled)
    y_te_p = clf.predict(x_test_scaled)
    
    tr_a = accuracy_score(y_train, y_tr_p)
    te_a = accuracy_score(y_test, y_te_p)
    pr = precision_score(y_test, y_te_p, average='binary')
    re = recall_score(y_test, y_te_p, average='binary')
    f1_m = f1_score(y_test, y_te_p, average='binary')
    
    cv_s = cross_val_score(clf, x_train_scaled, y_train, cv=skf, scoring='accuracy')
    
    final_results.append({
        'Model': name,
        'Accuracy': round(te_a, 4),
        'Precision': round(pr, 4),
        'Recall': round(re, 4),
        'F1-score': round(f1_m, 4),
        'CV Mean': round(cv_s.mean(), 4),
        'CV Std': round(cv_s.std(), 4),
        'Train Score': round(tr_a, 4),
        'Test Score': round(te_a, 4),
        'Fit Status': get_fit_status(tr_a, te_a)
    })

final_comparison_df = pd.DataFrame(final_results).sort_values(by='Accuracy', ascending=False).reset_index(drop=True)
display(final_comparison_df)

# Print Factual Summary
best_model_row = final_comparison_df.iloc[0]
print("=================== FACTUAL SUMMARY ===================")
print(f"1. Total models evaluated: {len(final_comparison_df)}")
print(f"2. Model with highest Test Accuracy: {best_model_row['Model']} ({best_model_row['Accuracy']*100:.2f}%)")
print(f"3. Model with highest F1-Score: {best_model_row['Model']} ({best_model_row['F1-score']:.4f})")
print(f"4. Model with highest Cross-Validation Mean Accuracy: {best_model_row['Model']} ({best_model_row['CV Mean']*100:.2f}% ± {best_model_row['CV Std']:.4f})")
print(f"5. Lowest Test Error Rate: {best_model_row['Model']} (Misclassification Rate: {(1 - best_model_row['Accuracy'])*100:.2f}%)")
print("=======================================================")
```
